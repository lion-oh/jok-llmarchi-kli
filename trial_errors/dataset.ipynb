```python
def preprocess_chat(conv_df:pd.DataFrame) -> str:

    result = []
    past_speaker = None
    for _, row in conv_df.iterrows():
        if past_speaker == row['speaker']:
            result += [' ' + row['utterance']]
        else:
            result += [f"\n\n {row['speaker']} : {row['utterance']}"]
            past_speaker = row['speaker']

    xdata = ''.join(result)
    return xdata
```
